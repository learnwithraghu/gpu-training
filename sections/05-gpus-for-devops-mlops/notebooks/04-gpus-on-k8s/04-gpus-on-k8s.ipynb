### Orchestrating GPUs

In Kubernetes, pods request GPU hardware using resource limits in their YAML configuration:

```yaml
resources:
  limits:
    nvidia.com/gpu: 1
```

The **NVIDIA Device Plugin** coordinates with the Kubernetes scheduler to assign physical GPU cards to these containers.

### GPU Slicing & MIG

For simple workloads like model inference, assigning a whole physical GPU (e.g. 16GB VRAM) to a single pod is expensive and inefficient.

**Multi-Instance GPU (MIG)** allows us to slice a single physical GPU into multiple independent virtual slices. This enables multiple pods to share the same physical card securely without interference.

### Step 1: Simulating GPU Pod Capacity

Let's write a small Python script to query the active GPU count. This simulates how the Kubernetes runtime detects assigned devices.

````python
import torch

gpu_count = torch.cuda.device_count()
print("Number of GPU devices detected by container:", gpu_count)
for i in range(gpu_count):
    print(f"Device {i}: {torch.cuda.get_device_name(i)}")
````

### Slicing Recommendations

* Use **MIG / GPU Slicing** for serving small, low-traffic models to maximize hardware utilization.
* Use **Dedicated GPUs** for training runs to avoid OOM limits from smaller VRAM slices.

### Module 4 Recap
