### The MLOps Incident Runbook

When serving or training models in production, three main issues arise:

1. **Unexplained Memory Consumption:** Zombie processes lock VRAM. Use `nvidia-smi` to find the culprit PIDs.
2. **Out of Memory (OOM):** The GPU cache is full of fragmented memory block gaps. Run `torch.cuda.empty_cache()` to reclaim them.
3. **Device Mismatch:** Tensors are stored on different devices. Ensure everything is aligned.

### Step 2: Resolving Out-Of-Memory (OOM) Gaps

When you delete variables in Python, the memory is marked as free in PyTorch but is not released back to the OS immediately. Let's see how to force cache release.

```python
import torch

x = torch.randn(5000, 5000, device="cuda")
del x  # Delete reference
torch.cuda.empty_cache()  # Free cached memory gaps
```

### Step 3: Avoiding Device Alignment Mismatch

Let's see what happens if we attempt to perform math between a CPU tensor and a GPU tensor.

```python
a = torch.randn(5, device="cpu")
b = torch.randn(5, device="cuda")
# This will raise a RuntimeError: Expected all tensors to be on the same device
try:
    c = a + b
except RuntimeError as e:
    print("Caught expected error:", e)
```
