### Managed Platforms vs. Bare VMs

* **AWS SageMaker & GCP Vertex AI:** Fully managed. They handle model provisioning, autoscaling endpoints, and security out-of-the-box. However, they are more expensive and experience slower boot times.
* **Raw VMs (EC2 / GCE):** You rent the virtual machine directly. You must manually install NVIDIA drivers, CUDA toolkits, and Docker. It is cheaper and boots instantly, but requires substantial operational overhead.

### Infrastructure Recommendations

* **Use Managed Endpoints (Vertex AI / SageMaker)** if you are serving mission-critical APIs with variable traffic and need automatic scaling.
* **Use Raw VMs** for batch offline ETL or during initial model experimentation where setup speed and cost are the primary constraints.

### Module 3 Recap

* **Managed cloud MLOps** handles scaling and security but costs more.
* **Raw VMs** offer total control and lower cost but require manual driver/docker management.
* Matching PyTorch CUDA compilations with the active GPU drivers prevents container crashes.
