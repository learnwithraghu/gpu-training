### The CPU Memory Pyramid

In computing, memory that is very fast is extremely expensive and small. Memory that is large is cheap but slow.

* **Registers:** Located directly inside the CPU calculation core. They are tiny (bytes) but operate at the speed of light.
* **Caches (L1/L2/L3):** Small storage pools (Megabytes) built into the CPU chip. They store active variables so the core doesn't have to wait for RAM.
* **System RAM:** The main memory (Gigabytes) connected via motherboard slots. It is far away from the CPU core.

* **Caches:** The drawers in your desk. Takes a few seconds to reach in and grab a folder.
* **System RAM:** The public library down the street. If you need a book, you have to pack your bags, walk down the street, and fetch it. Very slow!

### Step 1: Allocating Tensors in CPU RAM

Let's write a script to allocate a large tensor in System RAM and perform an operation on it, observing how CPU execution interacts with main memory.

```python
import time
import torch

# Allocate 20,000,000 floats in CPU System RAM
data = torch.randn(20000000, device="cpu")
t0 = time.perf_counter()  # Start CPU timer
result = data + 1.0  # Access memory and add 1.0
print(f"CPU RAM Access + Add: {time.perf_counter() - t0:.4f}s")
```

### Interpretation

Even though the math is extremely simple (just adding 1.0), the operation takes time because the CPU must fetch all 20,000,000 floats from the relatively slow System RAM. The bottleneck is not the addition; it's the memory travel time. In the next module, we look at the GPU side.
