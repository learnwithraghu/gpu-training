### The Reign of the CPU

For over 20 years, the Central Processing Unit (CPU) was the undisputed king of all computing.

Whether you were playing video games, running Excel, training a random forest model, or writing complex data pipelines, the CPU did it all.

Why? Because the CPU is designed to do any task you throw at it with incredible speed. It is optimized to run a single step as fast as physically possible.

### Analogy: The Formula 1 Race Car

Think of the CPU core as a **Formula 1 race car**.

It runs on a single-lane road at extreme speeds (3 to 5 GHz, which means 3 to 5 billion operations per second). However, it only has **one seat**—it carries one passenger (one task or thread) at a time.

### Step 1: Let's Import PyTorch

We will import PyTorch and check our processor speed. Let's run this import cell first.

```python
import time  # Timing measurements
import torch  # PyTorch tensor framework
print("PyTorch version:", torch.__version__)
```

### Step 2: Running Math on the CPU

Let's run a math operation on the CPU. We will create two vectors of size 1,000,000 and multiply them together element-by-element.

```python
# Create two float vectors on the CPU
a = torch.randn(1000000, device="cpu")
b = torch.randn(1000000, device="cpu")
t0 = time.perf_counter()  # Start CPU timer
c = a * b  # Element-wise multiply
print(f"CPU Time: {time.perf_counter() - t0:.6f}s")
```

### Interpretation

The CPU handles this math very quickly. This is why for decades, machine learning algorithms (like linear regression, decision trees, and pandas transformations) ran exclusively on the CPU.

But as dataset sizes grew, and deep learning emerged with billions of parameters, this single-car approach hit a wall. In the next module, we will meet the GPU.

### Module 1 Recap

* The **CPU** is a latency-optimized F1 race car designed for fast, sequential execution.
* For two decades, it did all data processing and machine learning.
