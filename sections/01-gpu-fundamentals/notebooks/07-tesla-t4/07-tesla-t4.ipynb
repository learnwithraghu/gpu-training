### Step 2: Running a Half-Precision Operation (FP16)

Tensor Cores are extremely fast when multiplying numbers represented in half-precision float format (`float16`). Let's run a matrix multiplication using `torch.float16` to leverage this hardware speed.

```python
# Create half-precision matrices directly on the GPU
a = torch.randn(2048, 2048, device="cuda", dtype=torch.float16)
b = torch.randn(2048, 2048, device="cuda", dtype=torch.float16)
torch.cuda.synchronize()
t0 = time.perf_counter()  # Start CPU timer
c = torch.matmul(a, b)  # Multiply matrices on GPU
torch.cuda.synchronize()  # Wait for completion
print(f"FP16 GPU Matmul Time: {time.perf_counter() - t0:.6f}s")
```

### Interpretation

That math took barely a few milliseconds! The Tesla T4's 320 Tensor Cores automatically detected the FP16 matrix operation and computed the multi-million multiplication steps instantly.
