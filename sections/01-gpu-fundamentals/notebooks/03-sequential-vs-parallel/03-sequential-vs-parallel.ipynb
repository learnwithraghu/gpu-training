### Analogy: Supermarket Lanes

Imagine a crowded supermarket with 1,000 customers waiting to pay:

* **Sequential (CPU style):** There is only **one** cashier. Even if the cashier is super-fast (GHz speed), every customer must wait for the person in front of them to finish. The queue takes time.
* **Parallel (GPU style):** The store opens **1,000 checkout lanes** simultaneously. Every customer walks to a counter and pays at the exact same time. The entire store is cleared instantly.

### Step 1: Let's Run Sequential Code on CPU

We will perform element-wise square operations. First, let's do it on the CPU.

```python
import time
import torch

x = torch.randn(5000000, device="cpu")
t0 = time.perf_counter()  # Start CPU timer
y = x ** 2  # Compute square sequentially
print(f"CPU sequential: {time.perf_counter() - t0:.6f}s")
```

### Step 2: Let's Run Parallel Code on GPU

Now, let's copy the same operation to the GPU (`device="cuda"`). Notice how we synchronize the timers to ensure accuracy.

```python
x_gpu = torch.randn(5000000, device="cuda")
torch.cuda.synchronize()  # Clean GPU stream queue
t0 = time.perf_counter()  # Start CPU timer
y_gpu = x_gpu ** 2  # Square in parallel
torch.cuda.synchronize()  # Wait for GPU completion
print(f"GPU parallel: {time.perf_counter() - t0:.6f}s")
```

### Interpretation

The GPU performs the operation almost instantly compared to the CPU. The GPU didn't step through the 5,000,000 numbers one-by-one; it distributed them across its thousands of cores and processed them all in parallel.

### Module 3 Recap

* **Sequential execution** processes tasks one-by-one (CPU checkout lane).
* **Parallel execution** processes tasks simultaneously (GPU checkout lanes).
* Vectorized tensor calculations compile into parallel operations on the GPU.
