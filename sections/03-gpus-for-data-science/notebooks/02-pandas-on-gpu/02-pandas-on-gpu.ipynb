### The Pandas Bottleneck

Pandas is the core tool for data manipulation in Python. However, Pandas runs operations sequentially on a single CPU core. 

When dealing with millions of rows, calculations like groupbys and mathematical column merges become extremely slow.

To solve this, Nvidia created **cuDF** (part of the RAPIDS ecosystem), which has the exact same API as Pandas but runs your commands in parallel on the GPU.

### Step 1: Converting Pandas to GPU Tensors

In practice, a simple way to speed up Pandas calculations is to extract the columns into PyTorch tensors and run them on the GPU. Let's create a Pandas dataframe.

```python
import pandas as pd
import torch

# Create a standard Pandas DataFrame on CPU
df = pd.DataFrame({"sales": [100.5, 200.2, 150.8, 300.1], "tax_rate": [0.1, 0.12, 0.08, 0.15]})
```

### Step 2: Run Math on the GPU

Now, let's extract the columns into GPU tensors and calculate the final tax amount in parallel.

```python
# Extract columns and load them directly onto the GPU
sales = torch.tensor(df["sales"].values, device="cuda")
tax_rate = torch.tensor(df["tax_rate"].values, device="cuda")
tax_amount = sales * tax_rate  # Multiplied in parallel on GPU
print("GPU calculated tax amounts:", tax_amount)
```

### Step 3: Write Back to Pandas

To put the results back into our CPU dataframe, we copy the calculated tensor back to the CPU.

```python
# Move the result back to CPU and add to Pandas DataFrame
df["tax_amount"] = tax_amount.to("cpu").numpy()
print(df)
```

### Module 2 Recap

* **Pandas** runs operations sequentially on a single CPU core.
* **cuDF** is the GPU-accelerated version of Pandas.
* You can easily speed up Pandas math by extracting columns into GPU PyTorch tensors and copying the results back.
