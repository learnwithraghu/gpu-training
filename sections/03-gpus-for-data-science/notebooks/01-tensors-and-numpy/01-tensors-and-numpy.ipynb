### Tensors vs. NumPy Arrays

If you have done data science in Python, you already know **NumPy**. NumPy arrays are excellent for holding tables and matrices of numbers.

However, NumPy was built before modern GPUs became popular. A NumPy array sits strictly in your computer's main memory (**System RAM**) and cannot run on a GPU.

**PyTorch Tensors** are the modern upgrade. They look and behave exactly like NumPy arrays, but they have a special property: **device awareness**. They can live on the CPU or be sent to the GPU.

### Step 1: Converting NumPy to PyTorch

Let's create a standard NumPy array and convert it into a PyTorch tensor. We use the `torch.from_numpy()` function to do this.

```python
import numpy as np
import torch

np_arr = np.array([10.0, 20.0, 30.0])  # Create NumPy array
tensor_arr = torch.from_numpy(np_arr)  # Convert to Tensor
print("PyTorch Tensor:", tensor_arr)
```

### Step 2: Checking Device Location

Let's check where our new tensor lives. Every tensor has a `.device` attribute that tells you where it is stored.

```python
# Check device location of the converted tensor
print("Tensor device location:", tensor_arr.device)
```

### Step 3: Moving to GPU VRAM

By default, converted tensors live on the CPU. To run operations on the GPU, we must send it to the CUDA device using the `.to("cuda")` method.

```python
# Send the tensor to the GPU memory
gpu_tensor = tensor_arr.to("cuda")
print("New tensor device:", gpu_tensor.device)
```

### Step 4: The NumPy Conversion Trap

What if we want to convert our GPU tensor back to a NumPy array? If we try calling `.numpy()` directly on a GPU tensor, it will throw a error because NumPy cannot access GPU VRAM. 

We must copy the data back to the CPU first.

```python
# Copy GPU tensor back to CPU before converting to NumPy
cpu_tensor = gpu_tensor.to("cpu")
numpy_back = cpu_tensor.numpy()
print("Back to NumPy array:", numpy_back)
```

### Module 1 Recap

* **NumPy arrays** are CPU-only and cannot run on a GPU.
* **PyTorch tensors** are device-aware (`device="cpu"` or `device="cuda"`).
* Use **`.to("cuda")`** to copy a tensor to the GPU.
* You must call **`.to("cpu")`** before converting a GPU tensor back to NumPy.
