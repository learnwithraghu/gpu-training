**Outcome:** learners define baseline, improvement, and evidence.

## Step 1 - Define the project in one line

Pause and ask each learner to choose a track: DevOps, Data Science, or Data Engineering.

```python
track = "data-science"
goal = "reduce training step time"
print(track, goal)
```

## Step 2 - Build baseline measurement

Expected: one numeric baseline metric.

```python
import time
import torch
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
x = torch.randn((1024, 1024), device=device)
y = torch.randn((1024, 1024), device=device)
```

```python
t0 = time.perf_counter()
_ = torch.matmul(x, y)
baseline = time.perf_counter() - t0
print(round(baseline, 4))
```

## Step 3 - Apply one change

Common mistake: changing many knobs at once.

```python
x_opt = x.half()
y_opt = y.half()
```

```python
t1 = time.perf_counter()
_ = torch.matmul(x_opt, y_opt)
improved = time.perf_counter() - t1
print(round(improved, 4))
```

## Step 4 - Report

Learners should explain both value and risk.

```python
delta = baseline - improved
print({"baseline": baseline, "improved": improved, "gain": delta})
```

## Final checkpoint

Ask learners:
- What changed?
- Why is this credible?
- What would you test next?
