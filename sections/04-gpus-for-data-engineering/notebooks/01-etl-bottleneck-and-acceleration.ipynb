# ETL Bottleneck and Acceleration

Trainer persona: we will profile one pipeline stage at a time.

**Outcome:** identify where acceleration is actually worth it.

## Step 1 - Small synthetic dataset

Ask learners: which ETL stage is usually slowest in your stack?

```python
import time
import torch
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
rows = torch.randn((200000, 8), device=device)
```

## Step 2 - Transform stage timing

Expected: arithmetic transforms should be vectorization-friendly.

```python
t0 = time.perf_counter()
rows = rows * 1.2 + 3.0
transform_time = time.perf_counter() - t0
```

```python
print("transform:", round(transform_time, 4))
```

## Step 3 - Aggregate stage timing

Common mistake: timing without matching input size between runs.

```python
t1 = time.perf_counter()
agg = rows.mean(dim=0)
agg_time = time.perf_counter() - t1
```

```python
print("aggregate:", round(agg_time, 4))
print(agg[:3])
```
