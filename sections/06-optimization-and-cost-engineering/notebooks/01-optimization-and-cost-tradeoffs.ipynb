# Optimization and Cost Tradeoffs

Trainer lens: faster does not always mean cheaper.

**Outcome:** compare two settings and estimate relative cost.

## Step 1 - Setup

Ask learners: what single metric would your manager care about first?

```python
import time
import torch
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
x = torch.randn((1024, 1024), device=device)
```

## Step 2 - Baseline run

Expected: one baseline time we can compare against.

```python
t0 = time.perf_counter()
_ = torch.relu(x)
base = time.perf_counter() - t0
```

```python
print("baseline:", round(base, 6))
```

## Step 3 - Candidate optimization

Common mistake: changing two variables at once.

```python
x16 = x.half()
t1 = time.perf_counter()
_ = torch.relu(x16)
```

```python
opt = time.perf_counter() - t1
print("optimized:", round(opt, 6))
```

```python
cost_index = opt / base if base else 0.0
print("relative_cost_index:", round(cost_index, 4))
```

## Checkpoint

Ask learners:
- Did speed improve enough to justify complexity?
- What risk did lower precision introduce?
