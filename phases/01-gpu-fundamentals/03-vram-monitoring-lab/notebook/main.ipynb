## Setup

```python
import torch
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
```

## Theory-to-code bridge
Allocated memory is actively used memory; reserved memory includes cached blocks.

```python
if not torch.cuda.is_available():
    raise RuntimeError('GPU runtime is required for this lab.')

def print_memory(tag):
    allocated = torch.cuda.memory_allocated() / (1024 ** 2)
    reserved = torch.cuda.memory_reserved() / (1024 ** 2)
    print(f'{tag} -> allocated: {allocated:.2f} MB, reserved: {reserved:.2f} MB')

print_memory('baseline')
```

## Benchmark or experiment

```python
buffers = []
for i, size in enumerate([2048, 4096, 8192], start=1):
    t = torch.randn(size, size, device='cuda')
    buffers.append(t)
    print_memory(f'after allocation {i} ({size}x{size})')

print('\nMemory summary snippet:')
print(torch.cuda.memory_summary()[:1200])
```

```python
del buffers
torch.cuda.empty_cache()
print_memory('after cleanup')
```

## Interpretation
Discuss the difference between allocated and reserved memory before and after cleanup.
